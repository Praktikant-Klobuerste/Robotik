```python
#kinematische Größen: Gelenkwinkel und Gliedlängen
q1,q2=symbols('theta_1 theta_2')
l1, l2, = symbols('l_1 l_2')
```

```python
# Wo ist KoordSys EE (Endeffektor) relativ zu KoordSys 1
TEE_1 = Matrix([[cos(q2), -sin(q2), l2*cos(q2)],[sin(q2), cos(q2), l2*sin(q2)] ,[0,0,1]])
```

```python
# Wo ist KoordSys 1 relativ zu KoordSys 0
T1_0 = Matrix([[cos(q1), -sin(q1), l1*cos(q1)],[sin(q1), cos(q1), l1*sin(q1)] ,[0,0,1]])
```

```python
# Vorwärtskinematik durch Verkettung der Einzeltransformationen
TEE_0 = simplify(T1_0 * TEE_1)
```

```python
# hier interessiert uns nur die Position des Endeffektors in der x,y-Ebene, also die ersten beiden Einträge der letzten Spalte
X = TEE_0[0:2,2]
```

**Differentielle Vorwärtskinematik:**  
aus $\vec x=\mathrm{f}(\vec\theta)$  
wird $\dot{\vec x} = \frac{\partial \mathrm{f}}{\partial\theta}\dot{\vec\theta}$  
mit $\frac{\partial \mathrm{f}}{\partial\theta} = J$ (Jacobi-Matrix)

```python
# Differentielle Vorwärtskinematik
J = X.jacobian((q1,q2))
```

```python
# mache eine ausführbare Funktion aus dem symbolischen J (die thetas werden zu Argumenten der Funktion, l1 und l2 werden mit Zahlenwerten ersetzt)
Jfun=lambdify((q1,q2),J.subs({l1:0.22,l2:0.18}))
```

```python
# Singulärwertzerlegung für verschiedene Konfigurationen anschauen
Grad=np.pi/180
Jn=Jfun(45*Grad,90*Grad)
```

```python
import numpy as np
u,s,v_trans=np.linalg.svd(Jn)
v=v_trans.T
```
