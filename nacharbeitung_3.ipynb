```python
import sympy as sp
```

```python
l1, l2, theta1, theta2 = sp.symbols('l_1, l_2, theta_1, theta_2')

sp.Matrix([l1, l2, theta1, theta2]).T
```

```python
TEE_1 = sp.Matrix([[sp.cos(theta2), -sp.sin(theta2), l2*sp.cos(theta2)],
                    [sp.sin(theta2), sp.cos(theta2), l2*sp.sin(theta2)],
                    [0, 0, 1]])

TEE_1
```

```python
T1_0 = sp.Matrix([[sp.cos(theta1), -sp.sin(theta1), l1*sp.cos(theta1)],
                    [sp.sin(theta1), sp.cos(theta1), l1*sp.sin(theta1)],
                    [0, 0, 1]])

T1_0
```

```python
TEE_0 = T1_0 * TEE_1 
TEE_0
```

```python
TEE_0 = sp.simplify(TEE_0)

TEE_0
```

```python
rEE_0 = TEE_0[0:2, 2]

rEE_0
```

```python
J = rEE_0.jacobian([theta1, theta2])
J
```

```python
x,y = rEE_0

sp.Matrix([[sp.diff(x, theta1), sp.diff(x, theta2)],
          [sp.diff(y, theta1), sp.diff(y, theta2)]])
```
