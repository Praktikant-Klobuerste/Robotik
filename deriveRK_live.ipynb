```python
from sympy import *
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "last_expr_or_assign"
```

```python
# notwendige symbolische Variablen anlegen...
(y0, f00, f10, f05, a0, a1, a2, a3, tau,h) = symbols("y0, f00, f10, f05, a0, a1, a2, a3, tau,h ")
# ... und anschauen
Matrix([tau,h,a0,a1,a2,a3,h]).T
```

```python
# Polynomansatz für ein Interval tau = 0 .. h
y = a0 + a1*tau + a2*tau**2 + a3*tau**3
# (1. Ableitung lassen wir berechnen)
yP = diff(y,tau)
```

```python
# Bedinungen an das Polynom in Form eines Gleichungssystems
eq = [y.subs(tau,0)-y0, # Wert links
      yP.subs(tau,0)-f00, # Steigung links
      yP.subs(tau,h)-f10, # Steigung rechts
      yP.subs(tau,h/2)-f05] # Steigung in der Mitte
```

```python
# Gleichungssystem lösen (lassen)
res=solve(eq,[a0,a1,a2,a3])
```

```python
# Lösung in Ansatz einsetzen und auswerten für tau = h (rechter Rand des Intervalls)
simplify(y.subs(tau,h).subs(res))
```
