```python
from sympy import *

omega_R, omega_L, v, psidot, J, R, d = symbols("omega_R, omega_L, v, psidot J, R, d")
Matrix([omega_R, omega_L, v, psidot, J, R, d]).T
```

```python
from sympy import *

v_vec, omega_vec = symbols("vvec, omegavec")

v_vec = Matrix([v, psidot])
omega_vec = Matrix([omega_L, omega_R])
omega_vec
```

```python
J = Matrix([[R/2,R/2], [-R/(2*d), R/(2*d)]])

J
```

```python
v_vec = J*omega_vec
v_vec
```

```python
werte = {
    omega_L: -4,
    omega_R: 4,
    R: Rational(1, 10),   # R = 0.1 m
    d: Rational(1, 2)     # d = 0.5 m
}

v_vec_eingesetzt = v_vec.subs(werte)
v_vec_eingesetzt
```
