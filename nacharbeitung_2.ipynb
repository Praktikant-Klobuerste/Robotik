```python
from sympy import *

omega_R, omega_L, v, psidot, R, a = symbols("omega_R, omega_L, v, psidot R, a")
Matrix([omega_R, omega_L, v, psidot, R, a]).T

```

```python
m, J_plat, J_R = symbols("m J_plat J_R")
```

```python
J = Matrix([[R/2, R/2], [-R/a, R/a]])
J
```

```python

omega_vec = Matrix([[omega_L], [omega_R]])
omega_vec
```

```python
v_vec = J * omega_vec
v_vec
```

```python
eq = Matrix([
    [v],
    [psidot]
]) - Matrix([
    [R/2, R/2],
    [-R/a, R/a]
]) * Matrix([
    [omega_L],
    [omega_R]
])

eq
```

```python
res = solve(
    list(eq),
    (omega_L, omega_R)
)

res[omega_L]
```

```python

# 4. Kinetische Energie
J_body = symbols("J_body")

E_kin = (
    m/2 * v**2
    + J_plat/2 * psidot**2
    + J_R/2 * (omega_L**2 + omega_R**2)
)

E_kin

```

```python
# 5. Raddrehzahlen durch ihre Lösungen ersetzen
E_kin = E_kin.subs(res)
E_kin
```

```python
# 6. Ergebnis vereinfachen
E_kin = simplify(E_kin)
E_kin
```

```python
E_pot = 0

L = E_kin-E_pot
L
```

```python
dl_dq_dot = Matrix([[diff(L, v)],
                [diff(L, psidot)]])

dl_dq_dot
```

```python
dl_dq_dot = simplify(dl_dq_dot)
dl_dq_dot
```
