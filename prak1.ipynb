```python
from sympy import *
import numpy as np 
import pandas as pd
import scipy
from matplotlib import pyplot as plt
```

```python
#y0=phi
#y1=x
#y2=y

def fun(t,y):
    dy= np.zeros(3)
    kappa= np.sqrt(y[1]**2+ y[2]**2)
    dy[0]= kappa
    dy[1]= np.cos(y[0])
    dy[2]= np.sin(y[0])
    return dy

#solve_ivp(fun,t_span,y_0, method, t_eval)
res= scipy.integrate.solve_ivp(fun, [0, 20], [0 ,0, 0], dense_output=True, method='RK45')

res2= res.sol(np.arange(0,20,0.01))

fig,ax= plt.subplots()
ax.plot(res2[1,:], res2[2,:])
```
